### Contextualização e Objetivo
Este notebook implementa a arquitetura **Lakehouse Medallion (Bronze -> Silver -> Gold)** utilizando **PySpark** e **Delta Lake** para o saneamento de dados de sensores industriais.

* **Desafio:** A base bruta possui 970 atributos e uma esparsidade extrema gerada por rotas distintas na linha de produção.
* **Solução:** Aplicar curadoria progressiva para eliminar atributos sem variabilidade e com alta taxa de dados ausentes, elevando a densidade informacional para consumo analítico.

```python
# 
# CAMADA BRONZE: INGESTÃO RAW & PERSISTÊNCIA EM DELTA LAKE (UNITY CATALOG)
#
# Objetivo: Realizar a ingestão imutável dos dados brutos do sensor, preservando
# a estrutura original (970 colunas) e registrando metadados de auditoria.
#

import builtins
import time
from pyspark.sql.functions import current_timestamp, lit

input_path = "/Volumes/workspace/default/pipeline_ufrj/train_numeric.csv"
bronze_delta_path = "/Volumes/workspace/default/pipeline_ufrj/lakehouse/bronze/bosch_numeric"

def run_bronze():
    t0 = time.time()
    df = (spark.read.option("header", "true").csv(input_path)
          .withColumn("timestamp_ingestao", current_timestamp())
          .withColumn("fonte_dados", lit("bosch_production_line")))
    df.write.format("delta").mode("overwrite").option("overwriteSchema", "true").save(bronze_delta_path)
    return time.time() - t0

print("🔥 Executando warm-up da Bronze...")
_ = run_bronze()

tempos_bronze = []
for i in range(3):
    t = run_bronze()
    tempos_bronze.append(t)
    print(f"  Execução {i+1}: {t:.2f}s")

tempo_ingestao_bronze = builtins.sum(tempos_bronze) / len(tempos_bronze)
print(f"✅ Camada Bronze — Média N=3: {tempo_ingestao_bronze:.2f}s")
```

```python
# 
# MENSURAÇÃO DO BASELINE DE QUALIDADE (CAMADA BRONZE COMPLETA)
# 
# Carrega a tabela Delta recém-criada no Volume
bronze_delta_path = (
    "/Volumes/workspace/default/pipeline_ufrj/lakehouse/bronze/bosch_numeric"
)
df_bronze = spark.read.format("delta").load(bronze_delta_path)

# Contagem de Registros e Dimensionalidade
total_linhas = df_bronze.count()
total_colunas_originais = (
    len(df_bronze.columns) - 2
)  # Descontando as 2 colunas de metadados[cite: 1]

print(f"📊 --- MÉTRICAS DE BASELINE DA BASE BRUTA ---")
print(f"• Total de Linhas Processadas: {total_linhas:,}")
print(f"• Total de Atributos Originais: {total_colunas_originais}")
print(
    f"• Total de Pontos de Observação (Matriz): {total_linhas * total_colunas_originais:,} células"
)
```

```python
from pyspark.sql.functions import col, count, when
import builtins
import time

import builtins
import time
from pyspark.sql.functions import col, count, when

try:
    df_bronze
except NameError:
    df_bronze = spark.read.format("delta").load(bronze_delta_path)

try:
    total_linhas
except NameError:
    total_linhas = df_bronze.count()

start_time_baseline = time.time()

cols_dados_bronze = [c for c in df_bronze.columns if c not in ["timestamp_ingestao", "fonte_dados"]]
total_colunas_bronze = len(cols_dados_bronze)

agg_baseline = [
    count(when(col(c).isNotNull(), col(c))).alias(f"{c}_non_nulls")
    for c in cols_dados_bronze
]
baseline_stats = df_bronze.select(agg_baseline).collect()[0].asDict()

total_valores_presentes_bronze = builtins.sum(
    baseline_stats[f"{c}_non_nulls"] for c in cols_dados_bronze
)
total_celulas_bronze = total_linhas * total_colunas_bronze
completude_baseline_bronze = (total_valores_presentes_bronze / total_celulas_bronze) * 100

end_time_baseline = time.time()

print("📊 --- BASELINE DE COMPLETUDE (CAMADA BRONZE BRUTA) ---")
print(f"• Total de Colunas Avaliadas: {total_colunas_bronze}")
print(f"• Total de Células Possíveis: {total_celulas_bronze:,}")
print(f"• Total de Valores Não Nulos: {total_valores_presentes_bronze:,}")
print(f"• Completude Média (Baseline Bronze): {completude_baseline_bronze:.2f}%")
print(f"• Tempo de Cálculo: {end_time_baseline - start_time_baseline:.2f}s")
```

```python
# 
# CAMADA SILVER: SANEAMENTO, CURATORIA E REDUÇÃO DE DIMENSIONALIDADE
# 
# Objetivo: Aplicar regras de curadoria técnica para eliminar colunas altamente
# esparsas (>95% nulos) e sem variabilidade (stddev = 0), preservando 100% dos registros.
# 

import builtins
import time
from pyspark.sql.functions import col, count, stddev, when

silver_delta_path = "/Volumes/workspace/default/pipeline_ufrj/lakehouse/silver/bosch_numeric"
threshold_nulos = 0.95

def run_silver():
    t0 = time.time()
    agg_stats = []
    for c in df_bronze.columns:
        if c not in ["timestamp_ingestao", "fonte_dados"]:
            agg_stats.append(count(when(col(c).isNotNull(), col(c))).alias(f"{c}_non_nulls"))
            agg_stats.append(stddev(col(c)).alias(f"{c}_stddev"))

    stats_row = df_bronze.select(agg_stats).collect()[0].asDict()

    cols_mantidas = []
    for c in df_bronze.columns:
        if c in ["timestamp_ingestao", "fonte_dados"]:
            continue
        non_nulls = stats_row[f"{c}_non_nulls"]
        std_val = stats_row[f"{c}_stddev"]
        tem_dados = (non_nulls / total_linhas) >= (1 - threshold_nulos)
        tem_variancia = (std_val is not None) and (std_val > 0)
        if c in ["Id", "Response"] or (tem_dados and tem_variancia):
            cols_mantidas.append(c)

    df_silver = df_bronze.select(cols_mantidas)
    df_silver.write.format("delta").mode("overwrite").option("overwriteSchema", "true").save(silver_delta_path)
    return time.time() - t0, len(cols_mantidas)

print("🔥 Executando warm-up da Silver...")
_ = run_silver()

tempos_silver = []
n_cols = None
for i in range(3):
    t, n_cols = run_silver()
    tempos_silver.append(t)
    print(f"  Execução {i+1}: {t:.2f}s")

tempo_silver = builtins.sum(tempos_silver) / len(tempos_silver)
print(f"✅ Camada Silver — Média N=3: {tempo_silver:.2f}s")
print(f"• Colunas Iniciais (Bronze): {total_colunas_originais}")
print(f"• Colunas Mantidas na Tabela Silver (com Id/Response): {n_cols}")
print(f"• Atributos de Sensor Mantidos (sem Id/Response): {n_cols - 2}")
print(f"• Redução de Dimensionalidade: {((total_colunas_originais - n_cols) / total_colunas_originais) * 100:.2f}%")
```

```python
# 
# CAMADA GOLD: CONSOLIDAÇÃO DE KPIS E ELEVAÇÃO DE DENSIDADE
# 
import builtins
import time
from pyspark.sql.functions import approx_count_distinct, col, count

silver_delta_path = "/Volumes/workspace/default/pipeline_ufrj/lakehouse/silver/bosch_numeric"
gold_delta_path = "/Volumes/workspace/default/pipeline_ufrj/lakehouse/gold/bosch_numeric"

def run_gold():
    t0 = time.time()
    df_silver_loaded = spark.read.format("delta").load(silver_delta_path)
    total_cols_silver = len(df_silver_loaded.columns)
    total_linhas_g = df_silver_loaded.count()

    agg_gold = []
    for c in df_silver_loaded.columns:
        if c not in ["timestamp_ingestao", "fonte_dados"]:
            agg_gold.append(count(col(c)).alias(f"{c}_count"))
            agg_gold.append(approx_count_distinct(col(c), rsd=0.05).alias(f"{c}_distinct"))

    gold_results = df_silver_loaded.select(agg_gold).collect()[0].asDict()

    total_celulas_possiveis = total_linhas_g * (total_cols_silver - 2)
    total_valores_presentes = builtins.sum(
        gold_results[f"{c}_count"] for c in df_silver_loaded.columns
        if c not in ["timestamp_ingestao", "fonte_dados"]
    )
    soma_cardinalidade = builtins.sum(
        gold_results[f"{c}_distinct"] for c in df_silver_loaded.columns
        if c not in ["timestamp_ingestao", "fonte_dados"]
    )

    completude = (total_valores_presentes / total_celulas_possiveis) * 100
    cardinalidade = soma_cardinalidade / (total_cols_silver - 2)

    df_silver_loaded.write.format("delta").mode("overwrite").option("overwriteSchema", "true").save(gold_delta_path)
    return time.time() - t0, completude, cardinalidade, total_linhas_g, total_cols_silver

print("🔥 Executando warm-up da Gold...")
_ = run_gold()

tempos_gold = []
completude = cardinalidade = total_linhas_final = total_cols_silver = None
for i in range(3):
    t, completude, cardinalidade, total_linhas_final, total_cols_silver = run_gold()
    tempos_gold.append(t)
    print(f"  Execução {i+1}: {t:.2f}s")

tempo_gold = builtins.sum(tempos_gold) / len(tempos_gold)
print(f"\n🏆 Camada Gold — Média N=3: {tempo_gold:.2f}s")
print(f"• Total de Linhas Preservadas: {total_linhas_final:,} (100% de retenção)")
print(f"• Atributos Mantidos: {total_cols_silver - 2} colunas")
print(f"• Densidade Informacional (Completude Média Mantida): {completude:.2f}%")
print(f"• Cardinalidade Média por Coluna: {cardinalidade:.2f} valores distintos")
```

```python
#Telemetria End-to-End
latencia_end_to_end = tempo_ingestao_bronze + tempo_silver + tempo_gold

tp_bronze = total_linhas / tempo_ingestao_bronze
tp_silver = total_linhas / tempo_silver
tp_gold = total_linhas / tempo_gold
tp_end_to_end = total_linhas / latencia_end_to_end

print('⏱️ --- TELEMETRIA COMPLETA DO PIPELINE (MÉDIA N=3) ---')
print(f'• Bronze:  {tempo_ingestao_bronze:.2f}s  |  Throughput: {tp_bronze:,.2f} reg/s')
print(f'• Silver:  {tempo_silver:.2f}s  |  Throughput: {tp_silver:,.2f} reg/s')
print(f'• Gold:    {tempo_gold:.2f}s  |  Throughput: {tp_gold:,.2f} reg/s')
print('------------------------------------------------------------------')
print(f'🚀 LATÊNCIA END-TO-END TOTAL: {latencia_end_to_end:.2f} segundos')
print(f'🚀 THROUGHPUT CONSOLIDADO (END-TO-END): {tp_end_to_end:,.2f} registros/segundo')
```

```python
# 
# CAMADA GOLD ENRIQUECIDA: ANÁLISE DE CONFORMIDADE E TAXA DE DEFEYTOS (RESPONSE)
# 
# Objetivo: Correlacionar os atributos saneados com a variável de negócio 'Response'
# (0 = Peça Conforme / 1 = Peça Defeituosa), entregando valor analítico direto
# para o controle de qualidade da fábrica.
# 

import time
from pyspark.sql.functions import avg, col, count, sum as spark_sum, when

start_time_gold_business = time.time()

gold_delta_path = "/Volumes/workspace/default/pipeline_ufrj/lakehouse/gold/bosch_numeric"
df_gold_loaded = spark.read.format("delta").load(gold_delta_path)

df_gold_clean = df_gold_loaded.withColumn("Response", col("Response").cast("int"))

kpi_response = df_gold_clean.groupBy("Response").agg(count("Id").alias("total_pecas"))

res_dict = {row["Response"]: row["total_pecas"] for row in kpi_response.collect()}
pecas_conformes = res_dict.get(0, 0)
pecas_defeituosas = res_dict.get(1, 0)
total_pecas_analisadas = pecas_conformes + pecas_defeituosas
taxa_defeito_pct = (pecas_defeituosas / total_pecas_analisadas) * 100

print("🏭 --- KPIS DE QUALIDADE E CONFORMIDADE INDUSTRIAL (CAMADA GOLD) ---")
print(f"• Total de Peças Inspecionadas: {total_pecas_analisadas:,}")
print(f"• Peças Conformes (Response = 0): {pecas_conformes:,} ({100 - taxa_defeito_pct:.2f}%)")
print(f"• Peças Defeituosas (Response = 1): {pecas_defeituosas:,} ({taxa_defeito_pct:.4f}%)")
print(f"• Taxa Global de Refugo/Falha da Linha: {taxa_defeito_pct:.4f}%")

gold_kpi_quality_path = "/Volumes/workspace/default/pipeline_ufrj/lakehouse/gold/kpi_qualidade_processo"
df_gold_clean.groupBy("Response").count().write.format("delta").mode("overwrite").save(gold_kpi_quality_path)

end_time_gb = time.time()
print(f"\n✅ Análise de Negócio Gold concluída em {end_time_gb - start_time_gold_business:.2f} segundos!")
```
